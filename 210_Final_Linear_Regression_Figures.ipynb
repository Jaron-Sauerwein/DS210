```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
cb = pd.read_csv('clean_bottles.csv')
```

```python
cb.plot.scatter('O2ml_L', 'T_degC');
```

```python
model = LinearRegression()

x_train = cb['O2ml_L'].values.reshape(-1, 1)
y_train = cb['T_degC']

model.fit(x_train, y_train)

y_pred = model.predict(cb['O2ml_L'].values.reshape(-1, 1))

cb.plot.scatter('O2ml_L', 'T_degC')
plt.plot(cb['O2ml_L'], y_pred, color='red', linewidth=3)
plt.xlabel('O2ml_L')
plt.ylabel('T_degC')
plt.show()
```

```python
print('Correlation coefficient: ', model.score(x_train, y_train))
```

```python
print(f'coefficient = {model.coef_[0]}')
print(f'intercept = {model.intercept_}')
```

```python
calc_pred = model.intercept_ + (x_train*model.coef_)
```

```python
(calc_pred == y_pred).all()
```

```python
correlation_matrix = cb.corr()
correlation_matrix
```

```python
new_data = pd.DataFrame({'O2ml_L': [3.3, 4.8, 6.4, 7.8]})

new_predictions = model.predict(new_data['O2ml_L'].values.reshape(-1, 1))

print(new_predictions)
```
