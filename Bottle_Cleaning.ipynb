```python
import pandas as pd
```

```python
bt = pd.read_csv('bottles.csv')
```

```python
def find_iqr_outliers(bt, column):
  Q1 = bt[column].quantile(0.25)
  Q3 = bt[column].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = bt[(bt[column] < lower_bound) | (bt[column] > upper_bound)]
  return outliers, lower_bound, upper_bound
```

```python
outliers, lower_bound, upper_bound = find_iqr_outliers(bt, 'Salnty')

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
print(f"Total Outliers Found: {len(outliers)}\n")

outliers
```

```python
outliers, lower_bound, upper_bound = find_iqr_outliers(bt, 'Depthm')

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
print(f"Total Outliers Found: {len(outliers)}\n")

outliers
```

```python
outliers, lower_bound, upper_bound = find_iqr_outliers(bt, 'T_degC')

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
print(f"Total Outliers Found: {len(outliers)}\n")

outliers
```

```python
outliers, lower_bound, upper_bound = find_iqr_outliers(bt, 'O2ml_L')

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
print(f"Total Outliers Found: {len(outliers)}\n")

outliers
```

```python
outliers, lower_bound, upper_bound = find_iqr_outliers(bt, 'STheta')

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
print(f"Total Outliers Found: {len(outliers)}\n")

outliers
```

```python
num_cols = ['Depthm', 'T_degC', 'Salnty', 'O2ml_L', 'STheta']

bt_cleaner = bt.copy()

for col in num_cols:
    Q1 = bt_cleaner[col].quantile(0.25)
    Q3 = bt_cleaner[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    bt_cleaner = bt_cleaner[(bt_cleaner[col] >= lower) & (bt_cleaner[col] <= upper) | bt_cleaner[col].isna()]

print(f"Original rows: {len(bt)}")
print(f"Rows remaining after outliers removed: {len(bt_cleaner)}")
```

```python
missing_total = bt_cleaner.isna().sum().sum()
print(f"Total values missing: {missing_total}")
```

```python
bt_clean = bt_cleaner.dropna()
print("Check for missing values:", bt_clean.isna().sum().sum())
```
